#### Processing and density computation

```python
cytoband_read = r"C:\Users\ext.andre.santos\Documents\Privacy_sens_human_genome\A_map_of_privacy-sensitivity_in_human_genomes-main\cytobands\cytobandFiltered_processed_str_dr_dg_gv_sorted.txt"
cb_df = pd.read_csv(cytoband_read, sep="\t")
cb_df
```

```python
def calc_AlzheimersSNPDensity(cytobands_df, alzheimer_df, col_name = "alzheimer_SNP_density"):
    densities = []
    for chrom in cytobands_df["Chromossome"].unique():             
        print("searching cytobands in chromossome: ", chrom)
        chr_cytobands = cytobands_df.loc[cytobands_df['Chromossome'] == chrom]
        chr_size = chr_cytobands["Last_index"][chr_cytobands.index[-1]]+1
        cytobands_alzheimers = np.zeros((chr_size), dtype=int)
        chr_alzheimers = alzheimer_df.loc[alzheimer_df["CHR_ID"] == chrom.split("r")[1]]
        #percorrer o dataset pelas vulnerabilidades no cromossoma e anotar na lista np.zeroes
        for row in chr_alzheimers.index:
            cytobands_alzheimers[int(chr_alzheimers["CHR_POS"][row])]+=1
        #Calculo da densidade para cada cytoband
        for cytoband in chr_cytobands.index:
            cytoband_array = cytobands_alzheimers[int(chr_cytobands["First_index"][cytoband]):int(chr_cytobands["Last_index"][cytoband])+1]
            dr_length = np.count_nonzero(cytoband_array)
            cytoband_length = chr_cytobands["Size"][cytoband]
            density = dr_length/cytoband_length
            densities.append(round(density, 10))
        
    cytobands_df.insert(len(cytobands_df.columns), col_name, densities, True)
```

```python
def writetxt(txt_path, df):
    with open(txt_path, 'w') as f:
        df_string = df.to_string(header=False, index=False)
        f.write(df_string)
        f.close

def writeNewtxt(file_path, df):
    """Writes a pandas dataframe into txt, columns separated by tabs ("\t")
    """
    f = open(file_path, 'w')
    str_cols = ""
    for col in df.columns[:-1]:
        str_cols += col + "\t"
    str_cols += df.columns.values[-1]+ "\n"
    f.write(str(str_cols))
    for row in df.values:
        str_row = ""
        for col in row[:-1]:
            str_row += str(col) + "\t"
        str_row += str(row[-1])+ "\n"
        f.write(str_row)
    f.close
```

```python
cytoband_read1 = r"C:\Users\ext.andre.santos\Documents\Privacy_sens_human_genome\A_map_of_privacy-sensitivity_in_human_genomes-main\cytobands\cytobandFiltered_processed_all_sorted.txt"
cb_df1 = pd.read_csv(cytoband_read1, sep="\t")
cb_df1
```

```python
def test_Alzheimersvalues(cytobands_df, col_name = "Alzheimer_metric"):
    alzheimer_densities = []
    metric_values = []
    for cytoband in cytobands_df.index:
        SNP_dens = cytobands_df["SNP_density"][cytoband]
        Alzheimer_snp_dens = cytobands_df["alzheimer_SNP_density"][cytoband]
        Alzheimer_dens = Alzheimer_snp_dens/SNP_dens
        alzheimer_densities.append(round(Alzheimer_dens, 10))
        #metric = 0.20 * SNP_dens + Alzheimer_dens * SNP_dens * 1000
        metric = 0.20 * SNP_dens + Alzheimer_snp_dens * 1000
        #metric = 0.20 * SNP_dens + Alzheimer_snp_dens * 1000 * SNP_dens 
        metric_values.append(round(metric, 10))

    cytobands_df.insert(len(cytobands_df.columns), "Alzheimer_dens", alzheimer_densities, True)
    cytobands_df.insert(len(cytobands_df.columns), col_name, metric_values, True)
```
