```python
def writeNewtxt(file_path, df):
    """Writes a pandas dataframe into txt, columns separated by tabs ("\t")
    """
    f = open(file_path, 'w')
    df_cols = ""
    for col in df.columns[:-1]:
        df_cols += str(col) + "\t"
    df_cols += df.columns.values[-1]+ "\n"
    f.write(str(df_cols))
    for row in df.index:
        df_row = ""
        for col in df.columns[:-1]:
            df_row += str(df[col][row]) + "\t"
        df_row += str(df[df.columns[-1]][row])+ "\n"
        f.write(df_row)
    f.close
```
